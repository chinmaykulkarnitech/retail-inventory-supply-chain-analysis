# 03. Exploratory Data Analysis (EDA) - Sales & Revenue Trends
**Project**: Retail Inventory & Supply Chain Performance Analysis

### Key Insights Explored:
- Revenue trends by category, store location, and season.
- Top & bottom performing product SKUs.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_sales = pd.read_csv('../data/processed/fact_sales.csv')
df_products = pd.read_csv('../data/processed/dim_product.csv')
df_merged = df_sales.merge(df_products, on='product_id')

cat_summary = df_merged.groupby('category')['sales_amount'].sum().sort_values(ascending=False)
print('--- Total Revenue by Category ---')
print(cat_summary)
```
