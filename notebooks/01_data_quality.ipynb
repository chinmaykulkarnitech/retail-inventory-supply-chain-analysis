# 01. Data Quality Audit & Exploration
**Project**: Retail Inventory & Supply Chain Performance Analysis

### Objectives:
- Audit raw retail datasets for missing values, duplicates, date inconsistencies, and foreign key integrity.
- Document data quality edge cases and establish cleaning rules.

```python
import os
import pandas as pd
import numpy as np

raw_dir = '../data/raw'
df_sales = pd.read_csv(os.path.join(raw_dir, 'raw_sales.csv'))
df_inventory = pd.read_csv(os.path.join(raw_dir, 'raw_inventory.csv'))
df_products = pd.read_csv(os.path.join(raw_dir, 'raw_products.csv'))
print(f'Raw Sales Shape: {df_sales.shape}')
print(f'Raw Inventory Shape: {df_inventory.shape}')
print(f'Raw Products Shape: {df_products.shape}')
```

## Data Quality Inspection Results
1. **Duplicates**: Duplicate records found in sales transaction IDs.
2. **Missing Values**: Missing store IDs in sales records and missing product categories.
3. **Formatting**: Date separators mixed ('/' vs '-').

```python
print('--- Null Values Audit ---')
print(df_sales.isnull().sum())

print('--- Duplicate Sales Transactions ---')
print(f'Duplicate rows: {df_sales.duplicated().sum()}')
```
